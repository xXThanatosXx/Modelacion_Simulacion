# Método de Newton-Raphson en Python

**Módulo I — Modelación matemática y métodos de solución**

Este notebook convierte a Python el código original escrito para MATLAB/Octave. El propósito es encontrar una raíz de la ecuación no lineal

$$f(x)=e^x+x+1=0.$$

Una raíz es un valor de $x$ para el cual $f(x)=0$. El método de Newton-Raphson construye aproximaciones sucesivas mediante

$$x_{k+1}=x_k-\frac{f(x_k)}{f'(x_k)}.$$

## Resultados de aprendizaje

Al finalizar, el estudiante podrá:

- implementar un método numérico iterativo en Python;
- estimar una derivada mediante diferencias finitas;
- interpretar tolerancia, error, convergencia y número de iteraciones;
- reconocer la influencia del valor inicial sobre el resultado.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def funcion(x):
    """Función cuya raíz se desea calcular: exp(x) + x + 1."""
    return np.exp(x) + x + 1
```

## Implementación del algoritmo

Para conservar la idea del código original, la derivada se aproxima con una diferencia finita hacia adelante:

$$f'(x)\approx\frac{f(x+h)-f(x)}{h}.$$

La función recibe cualquier función de Python, el valor inicial, la tolerancia, el número máximo de iteraciones y el paso $h$. Además, almacena el historial para poder estudiar la convergencia.

```python
def newton_raphson(fun, x0, tolerancia=1e-10, max_iteraciones=100, h=1e-5):
    """Calcula una raíz mediante Newton-Raphson y derivada numérica."""
    x_actual = float(x0)
    historial = []

    for iteracion in range(1, max_iteraciones + 1):
        valor_funcion = fun(x_actual)
        derivada = (fun(x_actual + h) - valor_funcion) / h

        if abs(derivada) < 1e-14:
            raise ZeroDivisionError(
                'La derivada es demasiado cercana a cero; cambie el valor inicial.'
            )

        x_nuevo = x_actual - valor_funcion / derivada
        error = abs(x_nuevo - x_actual)
        historial.append({
            'iteración': iteracion,
            'x_actual': x_actual,
            'f(x_actual)': valor_funcion,
            'derivada_aprox': derivada,
            'x_nuevo': x_nuevo,
            'error_absoluto': error,
        })

        if error < tolerancia:
            return x_nuevo, pd.DataFrame(historial)

        x_actual = x_nuevo

    raise RuntimeError(
        f'No se alcanzó la tolerancia después de {max_iteraciones} iteraciones.'
    )
```

## Ejecución e interpretación

Se utiliza $x_0=-1$ como aproximación inicial. La tabla permite observar cómo disminuye el error hasta satisfacer la tolerancia.

```python
raiz, iteraciones = newton_raphson(funcion, x0=-1.0)
print(f'Raíz aproximada: {raiz:.12f}')
print(f'Comprobación f(raíz): {funcion(raiz):.3e}')
print(f'Iteraciones realizadas: {len(iteraciones)}')
iteraciones
```

```python
x = np.linspace(-3, 1, 500)
fig, ejes = plt.subplots(1, 2, figsize=(12, 4))

ejes[0].plot(x, funcion(x), label=r'$f(x)=e^x+x+1$')
ejes[0].axhline(0, color='black', linewidth=0.8)
ejes[0].scatter([raiz], [funcion(raiz)], color='crimson', zorder=3, label='Raíz')
ejes[0].set(xlabel='x', ylabel='f(x)', title='Función y raíz encontrada')
ejes[0].grid(alpha=0.3)
ejes[0].legend()

ejes[1].semilogy(iteraciones['iteración'], iteraciones['error_absoluto'], marker='o')
ejes[1].set(xlabel='Iteración', ylabel='Error absoluto', title='Convergencia del método')
ejes[1].grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
