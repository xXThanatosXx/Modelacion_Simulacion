# 1.1 Un primer problema de planificación de la producción

## Declaración del problema

Una empresa produce dos versiones de un producto. Cada versión está hecha de la misma materia prima que cuesta 10 por gramo, y cada versión requiere dos tipos diferentes de mano de obra especializada para terminar. $U$ es la versión de mayor precio del producto. $U$ se vende a 270 la unidad y requiere 10 gramos de materia prima, una hora de mano de obra tipo $A$, dos horas de mano de obra tipo $B$. Debido al precio más alto, la demanda del mercado de $U$ está limitada a 40 unidades por semana. $V$ es la versión de menor precio del producto con demanda ilimitada que se vende a 210 por unidad y requiere 9 gramos de materia prima, 1 hora de mano de obra tipo $A$ y 1 hora de mano de obra tipo $B$. Estos datos se resumen en la siguiente tabla:

<div align="center">

| Versión | Materia prima <br> requerida | Se requiere mano de obra A <br> | Se requiere mano de obra B <br> | Mercado <br> Demanda | Precio |
| :-: | :-: | :-: | :-: | :-: | :-: | 
| U | 10 gramos | 1 hora | 2 horas | $\leq$ 40 unidades | 270 |
| V |  9 gramos | 1 hora | 1 hora | ilimitado | 210 |

</div>

La producción semanal de la empresa está limitada por la disponibilidad de mano de obra y el inventario de materias primas. La materia prima tiene una vida útil de una semana y debe solicitarse con antelación. Cualquier materia prima que sobra al final de la semana se desecha. La siguiente tabla detalla el costo y disponibilidad de materia prima y mano de obra.

<div align="center">

| Recurso | Cantidad <br> Disponible | Costo | 
| :-: | :-: | :-: |
| Materia Prima | ilimitado | 10/g |
| Labor A | 80 horas/semana | 50 / hora |
| Trabajo B | 100 horas/semana | 40 / hora | 

</div>

La empresa quiere maximizar sus beneficios brutos. 

1. ¿Cuánta materia prima se debe pedir por adelantado para cada semana? 
2. ¿Cuántas unidades de $U$ y $V$ debería producir la empresa cada semana?

## Modelo matemático

El planteamiento del problema anterior describe un problema de optimización. Reformular el planteamiento del problema como un modelo matemático implica algunos elementos cruciales:

* **variables de decisión**,
* **expresiones**,
* **función objetivo**,
* **restricciones**.

El punto de partida es desarrollar un modelo matemático que consiste en enumerar las variables de decisión relevantes para el problema en cuestión. **Variables de decisión** son cantidades que se pueden modificar para lograr el resultado deseado.  Si bien algunas variables de decisión introducidas en esta etapa pueden resultar redundantes más adelante, el objetivo en este punto es crear una lista completa de variables que serán útiles para expresar el objetivo y las limitaciones del problema. 

Para el planteamiento de este problema, a continuación se enumeran las variables de decisión con símbolos, descripciones y cualquier límite inferior y superior que se conozca a partir de los datos del problema.

<div align="center">

| Variable de decisión | Descripción | límite inferior | límite superior |
| :-: | :--- | :-: | :-: |
| $x_M$ | cantidad de materia prima utilizada | 0 | - |
| $x_A$ | cantidad de mano de obra A utilizada | 0 | 80 |
| $x_B$ | cantidad de mano de obra B utilizada | 0 | 100 |
| $y_U$ | número de unidades $U$ a producir | 0 | 40 |
| $y_V$ | número de unidades $V$ al producto | 0 | - |

</div>

El siguiente paso es formular una **función objetivo** que describa cómo mediremos el valor de las soluciones candidatas al problema. En este caso, el valor de la solución se mide por el beneficio que se desea maximizar. 

$$\max\ \text{profit}$$

La ganancia, a su vez, es igual a la diferencia entre los ingresos y el costo de las operaciones: 

$$
\begin{aligned}
    \text{profit} & = \text{revenue} - \text{cost} \\
\end{aligned}
$$

Los ingresos y los costos son **expresiones** lineales que se pueden escribir en términos de variables de decisión.

$$
\begin{aligned}
    \text{revenue} & = 270 y_U + 210 y_V \\
    \text{cost} & = 10 x_M + 50 x_A + 40 x_B  \\
\end{aligned}
$$

Como se muestra aquí, una expresión es una combinación algebraica de variables a las que se puede hacer referencia por su nombre. Las expresiones son útiles cuando la misma combinación de variables aparece en varios lugares de un modelo o cuando es deseable dividir expresiones más largas en unidades más pequeñas. Aquí hemos creado expresiones para ingresos y costos que simplifican la función objetivo.

Las variables de decisión $y_U, y_V, x_M, x_A, x_B$ deben satisfacer las condiciones específicas dadas en el planteamiento del problema. **Restricciones** son relaciones matemáticas entre variables o expresiones de decisión, formuladas como igualdades o desigualdades. En este problema, para cada recurso existe una restricción lineal que limita la producción general:

$$
\begin{aligned}
    10 y_U + 9 y_V  & \leq x_M & & \text{(raw material)}\\
    1 y_U + 1 y_V & \leq x_A & &\text{(labor A)} \\
    2 y_U + 1 y_V & \leq x_B & & \text{(labor B)}\\
\end{aligned}
$$

Ahora estamos listos para formular el problema de optimización matemática completo de la siguiente manera canónica: primero establecemos la función objetivo que se va a maximizar (o minimizar), luego enumeramos todas las restricciones y, por último, enumeramos todas las variables de decisión y sus límites:$$
\begin{align}
\max \quad & 270 y_U + 210 y_V - 10 x_M - 50 x_A - 40 x_B \\
\text{such that = s.t.} \quad & 10 y_U + 9 y_V  \leq x_M \nonumber \\
 & 1 y_U + 1 y_V \leq x_A \nonumber \\
 & 2 y_U + 1 y_V \leq x_B \nonumber \\
 & 0 \leq x_M \nonumber \\
 & 0 \leq x_A \leq 80 \nonumber \\
 & 0 \leq x_B \leq 100 \nonumber \\
 & 0 \leq y_U \leq 40 \nonumber \\
 & 0 \leq y_V.\nonumber 
\end{align}
$$

Esto completa la descripción matemática de este ejemplo de un problema de planificación de la producción. 

En muchos libros de texto, es habitual enumerar las variables de decisión bajo el símbolo $\max$ para distinguir claramente entre las variables y cualquier otro parámetro (que también podría expresarse como símbolos o letras) en la formulación del problema. Sin embargo, en todo este sitio web nos atenemos a la convención de que las variables de decisión son sólo aquellas que establecen explícitamente sus dominios como parte de la formulación del problema.

Una **solución óptima** del problema es cualquier vector de variables de decisión que cumpla con las restricciones y logre el objetivo máximo/mínimo.

Sin embargo, incluso para un problema simple como éste, no está inmediatamente claro cuál es la solución óptima. Aquí es exactamente donde entran en juego los algoritmos de optimización matemática: son procedimientos genéricos que pueden encontrar las soluciones óptimas de los problemas siempre que estos problemas puedan formularse de forma estandarizada como se indicó anteriormente. 

Para un profesional, la optimización matemática a menudo se reduce a formular el problema como un modelo anterior y luego pasarlo a uno de los paquetes de software comercial o de código abierto que puede resolver dicho modelo independientemente de cuál fue la *historia* original detrás del modelo. Para ello, necesitamos una interfaz de comunicación entre los modelos y los algoritmos. En este sitio web, adoptamos una interfaz basada en Python que es el paquete de modelado ```Pyomo```.

El siguiente paso es crear el modelo Pyomo correspondiente, lo cual realizaremos en el siguiente [cuaderno](02-production-planning-basic.ipynb).
