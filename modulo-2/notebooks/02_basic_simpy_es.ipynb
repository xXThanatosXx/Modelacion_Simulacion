# 3. Un ejemplo: una muestra de llamada de atención urgente

Este estudio de caso utiliza un modelo simple de centro de llamadas telefónicas de atención de urgencia, similar al servicio NHS 111 del Reino Unido.  Para aprender `simpy`, primero construiremos un modelo muy simple. En nuestra primera versión de este modelo, las llamadas al centro llegan **de manera determinista**.  Por ahora ignoraremos los recursos y actividades en el modelo y simplemente modelaremos un proceso de llegada determinista.   Las unidades de tiempo de simulación están en minutos.  Supongamos que hay 60 nuevas personas que llaman por hora (un tiempo entre llegadas fijo de 1,0 por minuto).

## 4. Los componentes básicos del modelo

Para construir nuestro modelo, necesitaremos los siguientes componentes...

### 4.1 Un entorno SimPy

`simpy` tiene una visión del mundo basada en procesos.  Estos procesos tienen lugar en un entorno.  Puede crear un entorno con la siguiente línea de código:

```python
env = simpy.Environment()
```

### 4.2 Tiempos de espera de SimPy

Podemos introducir **retrasos** o **actividades** en un proceso.  Por ejemplo, podrían ser la duración de una estancia en una sala o la duración de una operación o, en este caso, un **retraso entre llegadas (tiempo entre llegadas)**. En `simpy` controlas esto con el siguiente método:

```python
env.timeout(1.0)
```

### 4.3 Generadores

Los eventos en DES se modelan y programan en `simpy` usando **generadores** de Python (es decir, son el "mecanismo de procesamiento de eventos"). Un generador es una función que se comporta como un iterador, lo que significa que puede producir una **secuencia de valores** cuando se itera.

Por ejemplo, a continuación se muestra una función de generador básica que genera una nueva llegada cada minuto. Toma el **entorno** como parámetro. Luego llama internamente al método `env.timeout()` en un bucle infinito.

```python
def arrivals_generator(env):
    while True:
        yield env.timeout(1.0)
```

### 4.4 Proceso y ejecución de SimPy

Una vez que hemos codificado la lógica del modelo y creado una instancia de entorno, quedan dos instrucciones restantes que debemos codificar.

1. Configure el generador como un **proceso SimPy** usando `env.process()`

```python
env.process(arrivals_generator(env))
```

2. Ejecute el entorno para una **duración de ejecución** especificada por el usuario usando `env.run()`

```python
env.run(until=25)
```

El método de ejecución maneja el bucle infinito que configuramos en `arrivals_generator`. El modelo de simulación tiene un concepto interno de tiempo.  Finalizará la ejecución cuando su reloj interno alcance las 25 unidades de tiempo.

## 5. Crea el modelo

**Ahora que hemos cubierto los componentes básicos, codifiquemos el modelo real.** Tiene sentido crear primero la lógica de nuestro modelo.  El siguiente código generará llegadas cada 60,0/100,0 minutos.  Tenga en cuenta que la función toma un objeto de entorno como parámetro.

````python
def arrivals_generator(env):
    '''
    Callers arrive with a fixed inter-arrival time of 1.0 minutes.

    Parameters:
    ------
    env: simpy.Environment
    '''
    
    # don't worry about the infinite while loop, simpy will
    # exit at the correct time.
    while True:
        
        # sample an inter-arrival time.
        inter_arrival_time = 1.0
        
        # we use the yield keyword instead of return
        yield env.timeout(inter_arrival_time)
        
        # print out the time of the arrival
        print(f'Call arrives at: {env.now}')
````

Ahora que tenemos nuestra función de generador, podemos configurar el entorno, procesar y ejecutar la llamada.  Crearemos un parámetro `RUN_LENGTH` que puede cambiar para ejecutar el modelo durante diferentes períodos de tiempo.

**Considere:** ¿Qué pasaría si configuramos `RUN_LENGTH` en 50?

````python
# model parameters
RUN_LENGTH = 25

# create the simpy environment object
env = simpy.Environment()

# tell simpy that the `arrivals_generator` is a process
env.process(arrivals_generator(env))

# run the simulation model
env.run(until=RUN_LENGTH)
print(f'end of run. simulation clock time = {env.now}')
````
