# Conceptos básicos: funciones lineales, no lineales, maximización y minimización

**Módulo I — Introducción a la modelación y optimización**

Este notebook explica cuatro conceptos fundamentales mediante ejemplos sencillos y gráficas:

1. ¿Qué es una función?
2. ¿Qué es una función lineal?
3. ¿Qué es una función no lineal?
4. ¿Qué significa maximizar o minimizar?

No se requiere experiencia previa en optimización.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. ¿Qué es una función?

Una **función** es una regla que recibe un valor de entrada y produce un valor de salida.

Podemos imaginarla como una máquina:

$$\text{entrada} \longrightarrow \boxed{\text{función}} \longrightarrow \text{salida}$$

Por ejemplo:

$$f(x)=2x+3$$

Si la entrada es $x=4$, la salida es:

$$f(4)=2(4)+3=11.$$

En un problema empresarial, la entrada podría ser la cantidad producida y la salida podría ser el costo, el ingreso o el beneficio.

```python
def f(x):
    return 2*x + 3

entradas = np.arange(0, 6)
tabla_funcion = pd.DataFrame({
    'Entrada x': entradas,
    'Salida f(x)': f(entradas),
})
tabla_funcion
```

## 2. Función lineal

Una función lineal utilizada en modelación tiene la forma:

$$f(x)=mx+b,$$

donde:

- $x$ es la variable de entrada;
- $m$ es la pendiente o cambio constante;
- $b$ es el valor inicial;
- la gráfica es una línea recta.

### Ejemplo: costo de producción

Una empresa paga un costo fijo de 100 y un costo variable de 20 por cada unidad producida:

$$C(x)=100+20x.$$

Cada unidad adicional aumenta el costo exactamente en 20. Ese cambio constante es la característica principal del modelo lineal.

```python
x = np.linspace(0, 20, 100)
costo_lineal = 100 + 20*x

plt.figure(figsize=(8, 4.5))
plt.plot(x, costo_lineal, linewidth=2.5, label=r'$C(x)=100+20x$')
plt.scatter([0], [100], color='crimson', zorder=3, label='Costo fijo')
plt.xlabel('Unidades producidas')
plt.ylabel('Costo')
plt.title('Ejemplo de función lineal')
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## 3. Función no lineal

Una función es **no lineal** cuando su cambio no es constante y su gráfica no es una línea recta. Puede contener, por ejemplo:

- potencias como $x^2$ o $x^3$;
- productos entre variables como $xy$;
- raíces, cocientes o valores absolutos;
- funciones exponenciales, logarítmicas o trigonométricas.

### Ejemplo: costo por sobrecarga

Supongamos que operar una máquina a niveles altos genera desgaste adicional:

$$C(x)=100+10x+2x^2.$$

El término $2x^2$ hace que el costo crezca cada vez más rápido. El costo de producir una unidad adicional ya no es constante.

```python
costo_no_lineal = 100 + 10*x + 2*x**2

plt.figure(figsize=(8, 4.5))
plt.plot(x, costo_lineal, label='Lineal: 100 + 20x', linewidth=2)
plt.plot(x, costo_no_lineal, label='No lineal: 100 + 10x + 2x²', linewidth=2)
plt.xlabel('Unidades producidas')
plt.ylabel('Costo')
plt.title('Comparación entre una función lineal y una no lineal')
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## 4. ¿Cómo reconocerlas?

| Función | Tipo | Razón |
|---|---|---|
| $f(x)=5x+2$ | Lineal | La variable tiene exponente 1 |
| $f(x,y)=3x+4y$ | Lineal | Las variables no se multiplican entre sí |
| $f(x)=x^2+3x$ | No lineal | Aparece $x^2$ |
| $f(x,y)=xy$ | No lineal | Las variables se multiplican |
| $f(x)=e^x$ | No lineal | Es una función exponencial |
| $f(x)=10/x$ | No lineal | La variable está en el denominador |

> En programación lineal, tanto la función objetivo como todas las restricciones deben ser lineales.

## 5. ¿Qué es maximizar?

**Maximizar** significa encontrar el valor de las variables que hace que una función alcance el mayor resultado posible dentro de unas condiciones.

Ejemplos empresariales:

- maximizar el beneficio;
- maximizar las ventas;
- maximizar la producción;
- maximizar el nivel de servicio.

### Ejemplo

El beneficio por producir $x$ unidades es:

$$B(x)=-2x^2+40x-50.$$

La función primero aumenta y después disminuye. Esto puede representar que producir demasiado genera horas extra, desperdicio o saturación. Queremos encontrar la parte más alta de la curva.

```python
produccion = np.linspace(0, 20, 401)
beneficio = -2*produccion**2 + 40*produccion - 50
indice_maximo = np.argmax(beneficio)
x_maximo = produccion[indice_maximo]
beneficio_maximo = beneficio[indice_maximo]

print(f'Producción que maximiza el beneficio: {x_maximo:.2f}')
print(f'Beneficio máximo: {beneficio_maximo:.2f}')

plt.figure(figsize=(8, 4.5))
plt.plot(produccion, beneficio, linewidth=2.5)
plt.scatter([x_maximo], [beneficio_maximo], color='crimson', s=80,
            label=f'Máximo: ({x_maximo:.0f}, {beneficio_maximo:.0f})')
plt.xlabel('Producción')
plt.ylabel('Beneficio')
plt.title('Maximización del beneficio')
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## 6. ¿Qué es minimizar?

**Minimizar** significa encontrar el valor de las variables que hace que una función alcance el menor resultado posible dentro de unas condiciones.

Ejemplos empresariales:

- minimizar costos;
- minimizar desperdicios;
- minimizar tiempos de espera;
- minimizar distancias de transporte.

### Ejemplo

El costo de operar una planta a un nivel $x$ es:

$$C(x)=(x-8)^2+100.$$

Queremos localizar el punto más bajo de la curva.

```python
nivel_operacion = np.linspace(0, 16, 321)
costo = (nivel_operacion - 8)**2 + 100
indice_minimo = np.argmin(costo)
x_minimo = nivel_operacion[indice_minimo]
costo_minimo = costo[indice_minimo]

print(f'Nivel que minimiza el costo: {x_minimo:.2f}')
print(f'Costo mínimo: {costo_minimo:.2f}')

plt.figure(figsize=(8, 4.5))
plt.plot(nivel_operacion, costo, linewidth=2.5, color='#F28E2B')
plt.scatter([x_minimo], [costo_minimo], color='crimson', s=80,
            label=f'Mínimo: ({x_minimo:.0f}, {costo_minimo:.0f})')
plt.xlabel('Nivel de operación')
plt.ylabel('Costo')
plt.title('Minimización del costo')
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## 7. Maximizar y minimizar con restricciones

En un problema real no podemos elegir cualquier valor. Existen restricciones como presupuesto, tiempo, materias primas o demanda.

Por ejemplo, una empresa obtiene 30 por cada unidad de $A$ y 20 por cada unidad de $B$:

$$\max Z=30A+20B.$$

Pero dispone solamente de 100 horas:

$$2A+B\leq100,$$

y no puede producir cantidades negativas:

$$A,B\geq0.$$

La optimización busca el mejor punto que además cumpla todas las restricciones. Un resultado muy rentable que usa más de 100 horas no es una solución válida.

## 8. Resumen

| Concepto | Pregunta principal | Representación común |
|---|---|---|
| Función lineal | ¿La salida cambia a una tasa constante? | Línea recta |
| Función no lineal | ¿La tasa de cambio varía? | Curva |
| Maximización | ¿Cuál es el mayor valor posible? | Beneficio, producción o servicio |
| Minimización | ¿Cuál es el menor valor posible? | Costo, tiempo, distancia o desperdicio |

Una función describe la relación entre variables. La optimización utiliza esa función como objetivo y busca su mejor valor sin incumplir las restricciones del problema.
